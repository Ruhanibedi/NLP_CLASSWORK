## Program 1: Simple Rule-Based Chatbot using NLTK

**Aim:** Implement a simple rule-based chatbot using NLTK that matches patterns and generates predefined responses.

**Algorithm:**
1. Import NLTK's `chat` and `reflections` utilities.
2. Define a list of `(pattern, response-list)` pairs.
3. Create a `Chat` object with the patterns and reflections.
4. Call `converse()` to start an interactive session.
5. The bot replies until the user types `'quit'`.

```python
# Program 1: Simple Rule-Based Chatbot using NLTK
import nltk
from nltk.chat.util import Chat, reflections

nltk.download('punkt', quiet=True)

pairs = [
    (r"hi|hello|hey",
     ["Hello! How can I help you?",
      "Hi there! What can I do for you?"]),

    (r"what is your name?",
     ["I am Jason, your virtual assistant.",
      "My name is Jason!"]),

    (r"how are you?",
     ["I'm doing great, thank you!",
      "I'm fine. How about you?"]),

    (r"what can you do\??",
     ["I can answer questions, have a conversation, and help you learn NLP!",
      "I'm a rule-based chatbot. Try asking me something!"]),

    (r"tell me about (.*)",
     ["Sure! %1 is a very interesting topic.",
      "I know a bit about %1! What would you like to know?"]),

    (r"quit|bye|exit",
     ["Goodbye! Have a great day!",
      "Bye! Come back soon."]),

    (r"(.*)",
     ["I'm not sure I understand. Could you rephrase?",
      "Interesting! Tell me more."]),
]

def chatbot():
    print("Chatbot: Hello! I am Jason. Type 'quit' to exit.")
    chat = Chat(pairs, reflections)
    chat.converse()

if __name__ == "__main__":
    chatbot()
```

## Program 2: AI Chatbot using Transformers (DialoGPT)

**Aim:** Build a generative AI chatbot using the pre-trained DialoGPT model that can maintain multi-turn conversations.

**Algorithm:**
1. Load `microsoft/DialoGPT-medium` tokenizer and model.
2. Encode user input and append to conversation history.
3. Generate a response using the model.
4. Decode and print the response.
5. Repeat for multiple turns.

```python
# Program 2: AI Chatbot using DialoGPT
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch

tokenizer = AutoTokenizer.from_pretrained("microsoft/DialoGPT-medium")
model = AutoModelForCausalLM.from_pretrained("microsoft/DialoGPT-medium")

chat_history_ids = None
print("DialoGPT Chatbot started. Type 'quit' to exit.\n")

for step in range(5):
    user_input = input("You: ")
    if user_input.lower() == 'quit':
        break
    new_input_ids = tokenizer.encode(user_input + tokenizer.eos_token, return_tensors='pt')
    bot_input_ids = torch.cat([chat_history_ids, new_input_ids], dim=-1) if chat_history_ids is not None else new_input_ids
    chat_history_ids = model.generate(bot_input_ids, max_length=1000, pad_token_id=tokenizer.eos_token_id)
    reply = tokenizer.decode(chat_history_ids[:, bot_input_ids.shape[-1]:][0], skip_special_tokens=True)
    print(f"Bot: {reply}")
```

## Program 3: Question Answering using BERT

**Aim:** Use a pre-trained BERT model to extract answers from a given context passage.

**Algorithm:**
1. Load `deepset/roberta-base-squad2` pipeline.
2. Provide a context paragraph and a question.
3. The model extracts the answer span from the context.
4. Display the answer with confidence score.

```python
# Program 3: Question Answering using BERT
from transformers import pipeline

qa_pipeline = pipeline("question-answering", model="deepset/roberta-base-squad2")

context = """
The Amazon rainforest, often referred to as the lungs of the Earth, produces 20% of the world's oxygen
and is home to 10% of all species on Earth. It spans over 5.5 million square kilometres across nine countries.
Deforestation poses a severe threat; since 1970, over 750,000 square km have been lost.
"""

questions = [
    "What percentage of the world's oxygen does the Amazon produce?",
    "How many countries does the Amazon span?",
    "How much area has been lost since 1970?"
]

for q in questions:
    result = qa_pipeline(question=q, context=context)
    print(f"Q: {q}")
    print(f"A: {result['answer']}  (score: {result['score']:.2f})\n")
```

## Program 4: Extractive Summarization using NLTK

**Aim:** Summarize a document by extracting the most important sentences using word frequency analysis.

**Algorithm:**
1. Tokenize text into sentences and words.
2. Remove stopwords and compute word frequencies.
3. Score each sentence based on word frequencies.
4. Select top N sentences as the summary.

```python
# Program 4: Extractive Summarization using NLTK
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import sent_tokenize, word_tokenize
import heapq

nltk.download('punkt', quiet=True)
nltk.download('stopwords', quiet=True)
nltk.download('punkt_tab', quiet=True)

text = """
Artificial intelligence is transforming virtually every sector of the global economy.
In healthcare, AI systems can detect diseases from medical images with accuracy rivalling experienced physicians.
In finance, algorithmic trading systems process millions of transactions per second.
Manufacturing has been revolutionised through predictive maintenance — AI monitors machinery in real time.
Education platforms use adaptive learning algorithms that personalise content to individual students.
Despite the immense benefits, concerns persist: job displacement, algorithmic bias, and data privacy.
Policymakers worldwide are racing to draft regulations that balance innovation with public safety.
"""

stop_words = set(stopwords.words('english'))
words = word_tokenize(text.lower())
freq = {}
for word in words:
    if word.isalpha() and word not in stop_words:
        freq[word] = freq.get(word, 0) + 1

sentences = sent_tokenize(text)
scores = {}
for sent in sentences:
    for word in word_tokenize(sent.lower()):
        if word in freq:
            scores[sent] = scores.get(sent, 0) + freq[word]

summary = heapq.nlargest(3, scores, key=scores.get)
print("=== Extractive Summary ===")
for s in summary:
    print("-", s.strip())
```

## Program 5: Abstractive Summarization using BART

**Aim:** Generate a human-like summary using Facebook's BART Seq2Seq model.

**Algorithm:**
1. Load `facebook/bart-large-cnn` summarization pipeline.
2. Pass the input text to the model.
3. The model generates a new summary (not just extracted sentences).
4. Display the generated summary.

```python
# Program 5: Abstractive Summarization using BART
from transformers import pipeline

summarizer = pipeline("summarization", model="facebook/bart-large-cnn")

text = """
Artificial intelligence is transforming virtually every sector of the global economy.
In healthcare, AI systems can detect diseases from medical images with accuracy rivalling experienced physicians.
In finance, algorithmic trading systems process millions of transactions per second, identifying patterns invisible to human analysts.
Manufacturing has been revolutionised through predictive maintenance — AI monitors machinery in real time and forecasts failures before they occur.
Education platforms use adaptive learning algorithms that personalise content to individual students pace and style.
Despite the immense benefits, concerns persist: job displacement, algorithmic bias, data privacy, and the concentration of AI capabilities in a handful of large corporations.
Policymakers worldwide are racing to draft regulations that balance innovation with public safety.
"""

result = summarizer(text, max_length=80, min_length=30, do_sample=False)
print("=== Abstractive Summary (BART) ===")
print(result[0]['summary_text'])
```

## Program 6: Machine Translation using Transformers (Opus-MT)

**Aim:** Translate English text to French using Helsinki-NLP's MarianMT model.

**Algorithm:**
1. Load `Helsinki-NLP/opus-mt-en-fr` model and tokenizer.
2. Tokenize the English input text.
3. Generate translated token ids using the model.
4. Decode and display the French translation.

```python
# Program 6: Machine Translation using MarianMT (Opus-MT)
from transformers import MarianMTModel, MarianTokenizer

model_name = "Helsinki-NLP/opus-mt-en-fr"
tokenizer = MarianTokenizer.from_pretrained(model_name)
model = MarianMTModel.from_pretrained(model_name)

texts = [
    "Machine learning is a branch of artificial intelligence.",
    "Natural language processing enables computers to understand human language.",
    "Deep learning models are trained on large datasets."
]

print("=== English to French Translation (MarianMT) ===\n")
for text in texts:
    tokens = tokenizer([text], return_tensors="pt", padding=True)
    translated = model.generate(**tokens)
    result = tokenizer.decode(translated[0], skip_special_tokens=True)
    print(f"EN: {text}")
    print(f"FR: {result}\n")
```

## Program 7: Machine Translation using googletrans (API-based)

**Aim:** Use the Google Translate API via the `googletrans` library to translate text into multiple languages.

**Algorithm:**
1. Import `Translator` from `googletrans`.
2. Create a `Translator` object.
3. Call `translate(text, dest=language_code)`.
4. Print the original and translated text with detected source language.

```python
# Program 7: Machine Translation using googletrans
from googletrans import Translator

translator = Translator()

text = "Natural language processing is a fascinating field of artificial intelligence."

languages = {
    'fr': 'French',
    'es': 'Spanish',
    'de': 'German',
    'ta': 'Tamil',
    'hi': 'Hindi',
    'ja': 'Japanese'
}

print("=== Google Translate API (googletrans) ===\n")
print(f"Original (EN): {text}\n")

for code, name in languages.items():
    result = translator.translate(text, dest=code)
    print(f"{name}: {result.text}")
```
